```python
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
from collections import Counter
from torch.utils.data import Dataset, DataLoader
from nltk.tokenize import word_tokenize
import nltk
import re
```

```python
file_paths = [
    "data/text-data/youtube/text_1.txt",
    "data/text-data/youtube/text_2.txt"
]

docs = []
for path in file_paths:
    with open(path, "r") as file:
        docs.append(file.read())
```

```python
def clean_text(text: str) -> str:
    text = text.lower()
    text = re.sub(r"[^a-z0-9\s]", "", text)
    return text
    
# Tokenize the cleaned text
tokens = word_tokenize(clean_text(text))
print(tokens)
```

```python
def build_vocab(list_of_words):
    # bulid vocab
    vocab = {'<PAD>': 0, '<UNK>': 1}

    for token in Counter(list_of_words).keys():
        if token not in vocab:
            vocab[token] = len(vocab)
    return vocab
```

```python
vocab = build_vocab(tokens)
vocab
```

```python
len(vocab)
```

```python
cleaned_text = clean_text(text)
```

```python
input_sentences = []

for i in cleaned_text.split('\n'):
    if len(i) > 0:
        input_sentences.append(i.strip())
```

```python
input_sentences
```

```python
def text_to_indices(sentence, vocab):

    numberical_sentence = []

    for token in sentence:
        if token in vocab:
            numberical_sentence.append(vocab[token])
        else:
            numberical_sentence.append(vocab['<UNK>'])
    return numberical_sentence
```

```python
input_sentences[0]
```

```python
input_numerical_sentences = []

for sentence in input_sentences:
    tokens = word_tokenize(sentence)
    input_numerical_sentences.append(text_to_indices(tokens,vocab))
```

```python
len(input_numerical_sentences)
```

```python
training_sequence = []

for encoded_sentence in input_numerical_sentences:
    # Loop through the sentence tokens to build incremental sequences
    for token_index in range(1, len(encoded_sentence)):
        sequence_prefix = encoded_sentence[:token_index + 1]
        training_sequence.append(sequence_prefix)
```

```python
training_sequence
```

```python
len_list = []

for i in training_sequence:
    len_list.append(len(i))
max(len_list)
```

```python
max_len = max(len_list)
padded_training_sequence = []

for sequence in training_sequence:
    seq_list = list(sequence) 
    padded_seq = [0] * (max_len - len(seq_list)) + seq_list
    padded_training_sequence.append(padded_seq)
```

```python
padded_training_sequence = torch.tensor(padded_training_sequence,dtype =torch.long)
```

```python
X = padded_training_sequence[:,:-1]
y = padded_training_sequence[:,-1]
```

```python
class CustomDataset(Dataset):

    def __init__(self,X,y):
        self.X = X
        self.y = y

    def __len__(self):
        return self.X.shape[0]

    def __getitem__(self,idx):
        return self.X[idx], self.y[idx]
```

```python
dataset = CustomDataset(X,y)
```

```python
len(dataset)
```

```python
dataloader = DataLoader(dataset,batch_size=32,num_workers=2, shuffle=True)
```

```python
# Model Buliding
```

```python
class LSTMModel(nn.Module):

    def __init__(self,vocab_size):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size,100,padding_idx=0)
        self.lstm = nn.LSTM(100,150,batch_first=True)
        self.fc = nn.Linear(150,vocab_size)

    def forward(self,x):
        embedded = self.embedding(x)
        intermediate_hidden_states , (final_hidden_state,final_cell_state) = self.lstm(embedded)
        output = self.fc(final_hidden_state.squeeze(0))
        return output
```

```python
model = LSTMModel(len(vocab))
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
model.to(device)
```

```python
epochs = 50
learning_rate = 0.001
criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.Adam(model.parameters() , lr=learning_rate)
```

```python
# traing loop

for epoch in range(epochs):
    total_error = 0

    for batch_x , batch_y in dataloader:

        batch_x , batch_y = batch_x.to(device) , batch_y.to(device)

        optimizer.zero_grad()

        output = model(batch_x)

        loss = criterion(output,batch_y)

        loss.backward()

        optimizer.step()

        total_error += loss.item()

    print(f"Epoch : {epoch + 1} ,Loss : {total_error :.4f}")
```

```python
def prediction(model , vocab , text):
    model.eval()

    tokenized_text = word_tokenize(text.lower())

    # text --> numberical index

    numberical_text = text_to_indices(tokenized_text,vocab)

    # padding
    padded_seq = [0] * (61 - len(numberical_text)) + numberical_text
    padded_text = torch.tensor(padded_seq,dtype=torch.long).unsqueeze(0)

    padded_text = padded_text.to(device)


    # send model

    output = model(padded_text)

    # predicted index
    value , index = torch.max(output,dim=1)

    # merge with text
    return text + " " + list(vocab.keys())[index]
```

```python
def calculate_accuracy(model,dataloader,device):
    model.eval()

    total = 0
    correct = 0

    with torch.no_grad():
        for batch_x,batch_y in dataloader:
            batch_x,batch_y = batch_x.to(device) , batch_y.to(device)

            # model
            output = model(batch_x)

            # get the predicted words index
            _ , prediction = torch.max(output,dim=1)

            # Compare with actual labels
            correct += (prediction == batch_y).sum().item()
            total += batch_y.size(0)
    accuracy = correct / total * 100
    return accuracy

accuracy = calculate_accuracy(model,dataloader,device)
print(f"Model Accuracy : {accuracy}")
```
