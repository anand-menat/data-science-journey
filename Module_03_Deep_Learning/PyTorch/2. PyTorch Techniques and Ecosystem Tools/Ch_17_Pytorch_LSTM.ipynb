```python
import string
from nltk.tokenize import word_tokenize

# Define the characters you want to remove
chars_to_remove = "().',"

# Remove them from the text
cleaned_text = text.lower().translate(str.maketrans("", "", chars_to_remove))

# Tokenize the cleaned text
tokens = word_tokenize(cleaned_text)
print(tokens)
```

```python
def build_vocab(list_of_words):
    # bulid vocab
    vocab = {'<UNK>':0}

    for token in Counter(tokens).keys():
        if token not in vocab:
            vocab[token] = len(vocab)
    return vocab
```

```python
vocab = build_vocab(tokens)
vocab
```

```python
len(vocab)
```

```python
input_sentences = []

for i in cleaned_text.split('\n'):
    if len(i) > 0:
        input_sentences.append(i.strip())
```

```python
input_sentences
```

```python
def text_to_indices(sentence, vocab):

    numberical_sentence = []

    for token in sentence:
        if token in vocab:
            numberical_sentence.append(vocab[token])
        else:
            numberical_sentence.append(vocab['<UNK>'])
    return numberical_sentence
```

```python
input_sentences[0]
```

```python
input_numerical_sentences = []

for sentence in input_sentences:
    input_numerical_sentences.append(text_to_indices(sentence,vocab))
```

```python
len(input_numerical_sentences)
```

```python
training_sequence = []

for encoded_sentence in input_numerical_sentences:
    # Loop through the sentence tokens to build incremental sequences
    for token_index in range(1, len(encoded_sentence)):
        sequence_prefix = encoded_sentence[:token_index + 1]
        training_sequence.append(sequence_prefix)
```
