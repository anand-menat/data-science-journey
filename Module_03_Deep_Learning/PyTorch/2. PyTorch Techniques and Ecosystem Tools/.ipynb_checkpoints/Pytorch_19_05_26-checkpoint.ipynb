```python
import albumentations as A
from albumentations.pytorch import ToTensorV2
import cv2
import numpy as np
import torch
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image
```

```python
image_path = './data/tiny-imagenet-200/train/cats/edgar-nKC772R_qog-unsplash.jpg' 

# 1 image to numpy arrys unit8
image = cv2.imread(image_path)

image = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)


# 2. define the augumentation
tranform = A.Compose([
    A.HorizontalFlip(p=0.5),
    A.RandomBrightnessContrast(p=0.8),
    A.Rotate(limit=45,p=1.0),
])


# 3. Apply the tranformation
# Albumentations ALWAYS expects a dict with key 'image'

result = tranform(image=image)
augmented = result['image']

# 4. Display the results
fig, axs = plt.subplots(1, 2, figsize=(8, 4))
axs[0].imshow(image);      axs[0].set_title('Original');   axs[0].axis('off')
axs[1].imshow(augmented);  axs[1].set_title('Augmented');  axs[1].axis('off')
plt.suptitle('Minimal Albumentations Pipeline', fontweight='bold')
plt.tight_layout()
plt.show()

print(f"Input  shape: {image.shape},  dtype: {image.dtype}")
print(f"Output shape: {augmented.shape}, dtype: {augmented.dtype}")
```

### 4. Advanced Composition: `OneOf`, `SomeOf`, `Sequential`, `RandomOrder`

Beyond `Compose`, Albumentations provides powerful composition primitives that let you build non-trivial probabilistic pipelines.

```python
def show_augmentation_grid(transform, image, title, n=8, cols=4):
    rows = (n + cols - 1) // cols
    fig, axs = plt.subplots(rows, cols, figsize=(cols * 3, rows * 3))
    axs = axs.flatten()

    for i in range(n):
        result = transform(image=image)
        aug = result['image']

        # Extract which transform was applied
        replay = result['replay']
        applied = "Orignal"

        for t in replay['transforms']:
            if t['applied']:
                applied = t['__class_fullname__'].split('.')[-1]

                # Handle OneOf case (nested transforms)
                if 'transforms' in t:
                    for sub_t in t['transforms']:
                        if sub_t['applied']:
                            applied = sub_t['__class_fullname__'].split('.')[-1]
                break

        axs[i].imshow(aug)
        axs[i].set_title(applied, fontsize=8)
        axs[i].axis('off')

    for i in range(n, len(axs)):
        axs[i].axis('off')

    plt.suptitle(title, fontsize=13, fontweight='bold')
    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.show()
```

```python
# pipline

production_pipeline = A.ReplayCompose([
    A.OneOf([
        A.GaussianBlur(blur_limit=(3, 9), p=1.0),
        A.MotionBlur(blur_limit=9, p=1.0),
        A.MedianBlur(blur_limit=7, p=1.0),
        A.Defocus(radius=(1, 5), alias_blur=(0.05, 0.3), p=1.0)
            ])
        ])

show_augmentation_grid(production_pipeline, image, 'Production Pipeline (OneOf + Compose)', n=4)
```

```python
def my_show_augmentation_grid(transform, image, title, n=8, cols=4):
    rows = (n + cols - 1) // cols
    fig, axs = plt.subplots(rows, cols, figsize=(cols * 3, rows * 3))
    axs = axs.flatten()

    for i in range(n):
        result = transform(image=image)
        aug = result['image']

        # Extract which transform was applied
        replay = result['replay']
        applied = "Orignal"

        print(replay)
        print("----")
        print(replay['transforms'])

        for t in replay['transforms']:
            if t['applied']:
                applied = t['__class_fullname__'].split('.')[-1]

                # Handle OneOf case (nested transforms)
                if 'transforms' in t:
                    for sub_t in t['transforms']:
                        if sub_t['applied']:
                            applied = sub_t['__class_fullname__'].split('.')[-1]
                break

        axs[i].imshow(aug)
        axs[i].set_title(applied, fontsize=8)
        axs[i].axis('off')

    for i in range(n, len(axs)):
        axs[i].axis('off')

    plt.suptitle(title, fontsize=13, fontweight='bold')
    plt.tight_layout(rect=[0, 0, 1, 0.95])
    plt.show()
```
