## Object Oriented Programming 

**1. Define a python class Person with instance object variables name and age. Set Instance object variables in __init__() method. Also define show() method to display name and age of a person.**

```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def show(self):
        return f"Name :- {self.name}\nAge  :- {self.age}"
P1 = Person("Anand",25)
print(P1.show())
```

**2. Define a class Circle with instance object variable radius. Provide setter and getter for radius. Also define getArea() and getCircumference() methods.**

```python
class Circle:
    def __init__(self,radius):
        self.__radius = radius
        self.__pi = 22/7

    def get_radius(self):
        return self.__radius
        
    def set_radius(self,radius):
        self.__radius = radius

    def getArea(self):
        return self.__pi * (self.__radius **2)

    def getCircumference(self):
        return  (2 * self.__pi * self.__radius)

C1 = Circle(10)
print(C1.get_radius())
C1.set_radius(20)
print(C1.get_radius())
print(C1.getArea())
print(C1.getCircumference())
```

**3. Define a class Rectangle with length and breadth as instance object variables. Provide setDimensions(), showDimensions() and getArea() method in it.**

```python
class Rectangle:
    def __init__(self, length, breadth):
        self.__length = length
        self.__breadth = breadth

    def setDimensions(self, length, breadth):
        self.__length = length
        self.__breadth = breadth

    def showDimensions(self):
        return f"Length  :- {self.__length}\nBreadth :- {self.__breadth}"

    def getArea(self):
        return self.__length * self.__breadth

R1 = Rectangle(10, 5)
print(R1.showDimensions())
print(f"Area    :- {R1.getArea()}")

print("\n--- Changing Dimensions ---")
R1.setDimensions(20, 10)
print(R1.showDimensions())
print(f"Area    :- {R1.getArea()}")
```

**4. Define a class Book with instance object variables bookid, title and price. Initialise them via __init__() method. Also define method to show book variables.**

```python
class Book:
    def __init__(self,bookid,title,price):
        self.__bookid = bookid
        self.__title = title
        self.__price = price
 
    def showBook(self):
        return f"Book Id  :- {self.__bookid}\nTitle    :- {self.__title}\nPrice    :- {self.__price}"

book_1 = Book(1,"The Rudest Book Ever",500)
print(book_1.showBook())
```

**5. Define a class Team with instance object variable a list of team member names. Provide methods to input member names and display member names.`**

```python
class Team:
    # We set a default of an empty list, just in case a team starts with 0 people
    def __init__(self, initial_members=[]):
        self.__teamMember = initial_members.copy() 

    def add_names(self, names):
        if type(names) == list:
            self.__teamMember.extend(names) 
        elif type(names) == str:
            self.__teamMember.append(names)
            
    def show_name(self):
        output = "Member Names:\n"
        for name in self.__teamMember:
            output += f"- {name}\n"
        return output


team_1 = Team(["Anand"])

team_1.add_names(["Priya", "Rahul"])

team_1.add_names("Vikram")

print(team_1.show_name())
```
