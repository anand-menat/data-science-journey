**Lists** are built directly into Python and can hold a mix of different data types (like numbers, strings, and objects all together). **Arrays** require you to import a module, and they can only hold one specific data type (like *only* integers), making them much more memory-efficient and faster for math.

### High-Level Comparison

| Feature | Python List | Python Array (`array` module) | NumPy Array (`numpy`) |
| --- | --- | --- | --- |
| **Import Required?** | No (Built-in) | Yes (`import array`) | Yes (`import numpy`) |
| **Data Types** | Mixed (Heterogeneous) | Strict single type (Homogeneous) | Strict single type (Homogeneous) |
| **Memory Efficiency** | Low (Heavy overhead) | High (Compact) | Very High (Highly optimized) |
| **Math Operations** | Slow (Requires loops) | Slow (Requires loops) | Extremely fast (Vectorized) |

### 1. Python Lists (`list`)

Lists are the default, go-to data structure in Python. They are incredibly flexible and easy to use.

* **Flexibility:** You can store anything in them, and you can mix types.
* **Dynamic:** They can grow and shrink in size easily using `.append()` or `.remove()`.
* **The Catch:** Because they are so flexible, Python has to do a lot of work under the hood to manage them. Each item in a list is stored as a separate object in memory, which takes up more space and slows things down if you have millions of items.

```python
# A single list can hold an integer, a string, a float, and a boolean
my_list = [10, "Hello", 3.14, True]

```

### 2. Python Arrays (`array` module)

Python does have a built-in `array` module, though it isn't used very often. It provides a C-style array that restricts you to a single data type.

* **Strictness:** When you create one, you must declare a "type code" (like `'i'` for integers or `'d'` for decimals). It will crash if you try to put the wrong data type inside.
* **Memory:** They are much leaner than lists because the data is stored in contiguous (side-by-side) blocks of memory.
* **The Catch:** They don't support large-scale mathematical operations natively.

```python
import array

# The 'i' stands for integer. This array can ONLY hold integers.
my_array = array.array('i', [1, 2, 3, 4]) 

```

### 3. NumPy Arrays (`numpy.array`)

When most Python programmers talk about "arrays," they are actually talking about **NumPy arrays**. NumPy is a third-party library that is the absolute gold standard for data science and engineering in Python.

* **Speed:** NumPy is written mostly in C. It performs mathematical operations (like matrix multiplication) orders of magnitude faster than standard Python lists.
* **Vectorization:** You can apply a math operation to the entire array at once without writing a `for` loop.
* **The Catch:** You have to install NumPy (`pip install numpy`), and like the standard array, it forces all data to be of the same type.

```python
import numpy as np

my_numpy_array = np.array([1, 2, 3, 4])

# You can multiply the entire array by 2 instantly (Vectorization)
# Output: [2, 4, 6, 8]
print(my_numpy_array * 2) 

```

### Which one should you use?

* Use **Lists** for 90% of everyday Python programming. If you are collecting user inputs, scraping web data, or just need a simple collection of items, lists are perfect.
* Use **NumPy Arrays** if you are doing data science, machine learning, heavy math, or working with massive datasets (like images or audio files).
* Ignore the built-in **`array` module** unless you are doing low-level systems programming where you need strict C-compatibility and memory control without installing third-party libraries.

**1. Sort array values**

* **Input:** `[8, 2, 10, 1, 5, 8]`
* **Expected Output:** `[1, 2, 5, 8, 8, 10]`

````python
import array

arr_1 = array.array('i',[8, 2, 10, 1, 5, 8])


for i in range(len(arr_1)):
    for j in range(len(arr_1)):
        if arr_1[i] < arr_1[j]:
            arr_1[i] , arr_1[j] = arr_1[j] , arr_1[i]
arr_1
````

**3. Calculate the average of elements of a list**

* **Input:** `[10, 20, 30, 40, 50]`
* **Expected Output:** `30.0`

````python
List_2 = [10, 20, 30, 40, 50]

List_Sum = 0
ele_count = 0

for i in List_2:
    List_Sum += i
    ele_count += 1
    
print(List_Sum/ele_count)
````

**4. Create a list of the first N prime numbers**

* **Input:** `N = 6`
* **Expected Output:** `[2, 3, 5, 7, 11, 13]`

````python
N = 6
start_num = 2
prime_list = []

while len(prime_list) < N:
    is_prime = True
    last_num = int(start_num ** 0.5)
    
    for i in range(2,last_num+1):
        if start_num % i == 0:
            is_prime = False
            break
    if is_prime == True:
        prime_list.append(start_num)
    start_num += 1
prime_list
````

**5. Create a list of the first N terms of a Fibonacci series**

* **Input:** `N = 8`
* **Expected Output:** `[0, 1, 1, 2, 3, 5, 8, 13]`

````python
N = 8
a = 0
b = 1

fib_List = []

for i in range(1,N+1):
    c = a + b
    fib_List.append(a)
    a , b  = c, a
fib_List
````
