## 2. What is the Importance of Structuring Data?

Raw data without structure is **useless**. Structuring data helps us:

- **Organize** data logically for easy access
- **Store** data efficiently in memory
- **Process** data faster with the right operations
- **Retrieve** specific information quickly

### Example: Searching in Unstructured vs Structured Data

1. Dictionary :- Finding words from Dictionary
2. Map :- Finding locations from Map
3. Ledger :- Tranking Profit and Loss from Ledger (Data,Time,Product,City)

````python
import time

# Unstructured: List (Linear Search) — O(n)
data_list = list(range(1, 100001))

start = time.perf_counter()
found = 99999 in data_list
end = time.perf_counter()
print(f"List search time   : {(end - start)*1e6:.2f} microseconds")

# Structured: Set (Hash-based Lookup) — O(1)
data_set = set(range(1, 100001))

start = time.perf_counter()
found = 99999 in data_set
end = time.perf_counter()
print(f"Set search time    : {(end - start)*1e6:.2f} microseconds")

print("\n Structured data (Set) is significantly faster")
````

## 4. Where Do Data Structures Reside?

Data structures live in your computer's **primary memory (RAM)** during program execution.

```
┌─────────────────────────────────────────┐
│              RAM (Memory)               │
│                                         │
│  ┌──────────┐      ┌──────────────────┐ │
│  │  STACK   │      │      HEAP        │ │
│  │          │      │                  │ │
│  │ - Local  │      │ - Dynamic alloc  │ │
│  │   vars   │      │ - Objects        │ │
│  │ - Fixed  │      │ - Linked Lists   │ │
│  │   size   │      │ - Trees, Graphs  │ │
│  └──────────┘      └──────────────────┘ │
└─────────────────────────────────────────┘
```

### Stack Memory vs Heap Memory

| Feature | Stack | Heap |
|---------|-------|------|
| **Allocation** | Compile time | Runtime |
| **Size** | Fixed / Limited | Flexible / Large |
| **Speed** | Very Fast | Slower |
| **Managed by** | OS automatically | Programmer / GC |
| **Stores** | Primitives, function calls | Objects, dynamic DS |
| **Examples** | `int x = 5` | `list = [1,2,3]` |

````python
import sys

# Primitive (Stack-like in memory)
x = 42
print(f"Integer size       : {sys.getsizeof(x)} bytes")
print(f"Memory address     : {hex(id(x))}")

# Dynamic DS (Heap)
my_list = [1, 2, 3, 4, 5]
print(f"\nList size          : {sys.getsizeof(my_list)} bytes")
print(f"Memory address     : {hex(id(my_list))}")

my_dict = {'a': 1, 'b': 2}
print(f"\nDict size          : {sys.getsizeof(my_dict)} bytes")
print(f"Memory address     : {hex(id(my_dict))}")
````

### 🔷 1. Array (List)
- Contiguous memory locations
- Fixed type elements (in low-level languages)
- **Access**: O(1) | **Search**: O(n) | **Insert/Delete**: O(n)

````python
# Array / List
arr = [10, 20, 30, 40, 50]
print("Array:", arr)
print("Access index 2:", arr[2])          # O(1)
arr.append(60)                            # O(1) amortized
arr.insert(2, 25)                         # O(n)
arr.remove(20)                            # O(n)
print("After operations:", arr)
````

### 🔷 2. String
- Immutable sequence of characters
- Supports slicing, searching, pattern matching

````python
# String
s = "Hello, DSA!"
print("String      :", s)
print("Reverse     :", s[::-1])
print("Uppercase   :", s.upper())
print("Substring   :", s[7:10])
print("Find 'DSA'  :", s.find('DSA'))
````

### 🔷 3. Set
- Unordered collection of **unique** elements
- Based on Hash Table — O(1) average operations

````python
# Set
s = {1, 2, 3, 4, 4, 5, 5}  # duplicates removed
print("Set:", s)
s.add(6)
s.discard(1)
print("After add(6), remove(1):", s)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print("Union       :", a | b)
print("Intersection:", a & b)
print("Difference  :", a - b)
````

### 🔷 4. Dictionary (Hash Map)
- Key-value pairs

````python
# Dictionary
student = {
    "name": "Arjun",
    "age": 20,
    "grade": "A"
}
print("Dict:", student)
print("Name:", student["name"])
student["score"] = 95
print("After adding score:", student)

# Frequency count — classic dictionary use case
text = "banana"
freq = {}
for char in text:
    freq[char] = freq.get(char, 0) + 1
print("\nChar frequency in 'banana':", freq)
````

### 🔷 5. Stack
- **LIFO** (Last In, First Out)
- Operations: push, pop, peek
- Use cases: undo/redo, function call stack, bracket matching

````python
# Stack using list
stack = []
stack.append(10)   # push
stack.append(20)
stack.append(30)
print("Stack       :", stack)
print("Top (peek)  :", stack[-1])
print("Pop         :", stack.pop())
print("Stack now   :", stack)

# Practical: Balanced Brackets
def is_balanced(expr):
    stk = []
    pairs = {')': '(', '}': '{', ']': '['}
    for ch in expr:
        if ch in '({[':
            stk.append(ch)
        elif ch in ')}]':
            if not stk or stk[-1] != pairs[ch]:
                return False
            stk.pop()
    return len(stk) == 0

print("\n'{[()]}' balanced:", is_balanced('{[()]}'))
print("'{[(])}' balanced:", is_balanced('{[(])}'))
````

### 🔷 7. Linked List
- Nodes connected via pointers
- Dynamic size — no memory waste
- **Insert/Delete at head**: O(1) | **Search**: O(n)

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        curr = self.head
        while curr.next:
            curr = curr.next
        curr.next = new_node

    def display(self):
        nodes = []
        curr = self.head
        while curr:
            nodes.append(str(curr.data))
            curr = curr.next
        print(" -> ".join(nodes) + " -> None")

ll = LinkedList()
for val in [10, 20, 30, 40]:
    ll.append(val)

print("Linked List:")
ll.display()
````

### 🔷 8. Tree
- Hierarchical, non-linear structure
- Root → Parent → Children → Leaf
- **Binary Search Tree**: O(log n) for search, insert, delete (balanced)

````python
class TreeNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

def insert_bst(root, val):
    if not root:
        return TreeNode(val)
    if val < root.val:
        root.left = insert_bst(root.left, val)
    else:
        root.right = insert_bst(root.right, val)
    return root

def inorder(root):
    if root:
        inorder(root.left)
        print(root.val, end=' ')
        inorder(root.right)

root = None
for val in [5, 3, 7, 1, 4, 6, 8]:
    root = insert_bst(root, val)

print("BST Inorder (sorted):", end=' ')
inorder(root)
````

### 🔷 9. Graph
- Nodes (vertices) connected by edges
- Can be directed/undirected, weighted/unweighted
- Use cases: maps, social networks, web crawling

````python
from collections import defaultdict, deque

# Graph as adjacency list
graph = defaultdict(list)
edges = [(0,1),(0,2),(1,3),(2,4),(3,5)]
for u, v in edges:
    graph[u].append(v)
    graph[v].append(u)

print("Adjacency List:")
for node, neighbors in graph.items():
    print(f"  {node} → {neighbors}")

# BFS Traversal
def bfs(start):
    visited = set()
    queue = deque([start])
    result = []
    while queue:
        node = queue.popleft()
        if node not in visited:
            visited.add(node)
            result.append(node)
            queue.extend(graph[node])
    return result

print("\nBFS from node 0:", bfs(0))
````

## 6. Algorithms

> An **Algorithm** is a step-by-step procedure to solve a problem.

### Algorithm Categories

| # | Algorithm Type | Key Idea | Common Problems |
|---|---------------|----------|-----------------|
| 1 | **Two Pointers** | Use 2 pointers to reduce O(n²) to O(n) | Pair sum, reverse string |
| 2 | **Sliding Window** | Maintain a window of fixed/variable size | Max subarray, longest substring |
| 3 | **Searching** | Find an element efficiently | Binary search |
| 4 | **Sorting** | Arrange elements in order | Merge sort, Quick sort |
| 5 | **Recursion** | Function calls itself | Factorial, Fibonacci, DFS |
| 6 | **Backtracking** | Try all paths, undo on failure | N-Queens, Sudoku |
| 7 | **Greedy** | Pick the best local choice | Activity selection, Dijkstra |
| 8 | **Dynamic Programming** | Cache overlapping subproblems | Knapsack, LCS, Coin change |

### Quick Demos

````python
# ---- Two Pointers: Find pair with target sum ----
def two_sum_sorted(arr, target):
    left, right = 0, len(arr) - 1
    while left < right:
        s = arr[left] + arr[right]
        if s == target:
            return (arr[left], arr[right])
        elif s < target:
            left += 1
        else:
            right -= 1
    return None

arr = [1, 3, 5, 7, 9, 11]
print("Two Pointers — Pair with sum 14:", two_sum_sorted(arr, 14))
````

````python
# ---- Sliding Window: Max sum of k consecutive elements ----
def max_sum_window(arr, k):
    window_sum = sum(arr[:k])
    max_sum = window_sum
    for i in range(k, len(arr)):
        window_sum += arr[i] - arr[i - k]
        max_sum = max(max_sum, window_sum)
    return max_sum

arr = [2, 1, 5, 1, 3, 2]
print("Sliding Window — Max sum of 3 elements:", max_sum_window(arr, 3))
````

````python
# ---- Binary Search ----
def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

arr = [1, 3, 5, 7, 9, 11, 13, 15]
print("Binary Search — index of 9:", binary_search(arr, 9))
````

````python
# ---- Recursion: Fibonacci ----
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print("Fibonacci(0-9):", [fibonacci(i) for i in range(10)])
````

````python
# ---- Sorting: Merge Sort ----
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)

def merge(left, right):
    result, i, j = [], 0, 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i]); i += 1
        else:
            result.append(right[j]); j += 1
    return result + left[i:] + right[j:]

arr = [38, 27, 43, 3, 9, 82, 10]
print("Merge Sort:", merge_sort(arr))
````

````python
# ---- Dynamic Programming: Fibonacci with Memoization ----
from functools import lru_cache

@lru_cache(maxsize=None)
def fib_dp(n):
    if n <= 1:
        return n
    return fib_dp(n - 1) + fib_dp(n - 2)

print("DP Fibonacci(50):", fib_dp(50))   # instant with memoization
````

## 7️. Time & Space Complexity

> **Complexity** measures how an algorithm's resource usage grows with input size.

### Big O Notation — Common Complexities

| Big O | Name | Example |
|-------|------|---------|
| O(1) | Constant | Array index access |
| O(log n) | Logarithmic | Binary search |
| O(n) | Linear | Loop through array |
| O(n log n) | Linearithmic | Merge sort |
| O(n²) | Quadratic | Nested loops |
| O(2ⁿ) | Exponential | Recursive Fibonacci |
| O(n!) | Factorial | Permutations |

**Best to Worst:** O(1) < O(log n) < O(n) < O(n log n) < O(n²) < O(2ⁿ) < O(n!)

````python
import time
import math

def measure(func, *args):
    start = time.perf_counter()
    result = func(*args)
    end = time.perf_counter()
    return (end - start) * 1e6  # microseconds

n = 10000
arr = list(range(n))

# O(1)
t1 = measure(lambda: arr[n // 2])

# O(log n)
t2 = measure(binary_search, arr, n - 1)

# O(n)
t3 = measure(lambda: sum(arr))

# O(n log n)
t4 = measure(merge_sort, arr[:1000])

# O(n²)
t5 = measure(lambda: [[i*j for j in range(100)] for i in range(100)])

print(f"O(1)      : {t1:.2f} μs")
print(f"O(log n)  : {t2:.2f} μs")
print(f"O(n)      : {t3:.2f} μs")
print(f"O(n log n): {t4:.2f} μs (n=1000)")
print(f"O(n²)     : {t5:.2f} μs (n=100)")
````

### Space Complexity Examples

````python
# O(1) space — no extra memory
def sum_array(arr):
    total = 0           # just 1 variable
    for x in arr:
        total += x
    return total

# O(n) space — stores all elements
def copy_array(arr):
    return [x * 2 for x in arr]   # new list of size n

# O(n) space — recursion stack
def recursive_sum(arr, i=0):
    if i == len(arr):
        return 0
    return arr[i] + recursive_sum(arr, i + 1)

import sys
test = list(range(100))
print("O(1) space result  :", sum_array(test))
print("O(n) space result  :", copy_array([1,2,3]))
print("Recursion result   :", recursive_sum(test))
````
