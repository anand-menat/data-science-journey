```python
from manim import *

class VectorAddition(Scene):
    def construct(self):
        plane = NumberPlane()
        self.add(plane)

        v1 = Vector([2,3], color=BLUE)
        v2 = Vector([3,1], color=GREEN).shift(v1.get_end())

        self.play(GrowArrow(v1))
        self.play(GrowArrow(v2))
```
