```python
dice_1 = range(1, 7)
dice_2 = range(1, 7)

sum_pairs = {}

for d1 in dice_1:
    for d2 in dice_2:
        s = d1 + d2
        
        if s not in sum_pairs:
            sum_pairs[s] = []
        
        sum_pairs[s].append((d1, d2))
```

```python
for i,j in sum_pairs.items():
    print(i,":",j)
```

```python
from collections import defaultdict

sum_pairs = defaultdict(list)

for d1 in range(1, 7):
    for d2 in range(d1, 7): 
        s = d1 + d2
        sum_pairs[s].append((d1, d2))

for i,j in sum_pairs.items():
    print(i,":",j)
```

```python
# total pair caluations of pairs
```

```python
sample_out = []
for i in sum_pairs:
    for j in sum_pairs[i]:
        sample_out.append(j)
```
