```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
```

```python
# 1. Set seaborn theme
sns.set_theme(style="whitegrid")

# 2. Generate sample data (1000 data points each)
np.random.seed(42)

# Normal Distribution (The Baseline)
data_normal = np.random.normal(loc=0, scale=1, size=1000)

# Peaked in Middle / Heavy-Tailed (Leptokurtic) - using Laplace distribution
data_peaked = np.random.laplace(loc=0, scale=1, size=1000)

# Flat in Middle / Light-Tailed (Platykurtic) - using Uniform distribution
data_flat = np.random.uniform(low=-3, high=3, size=1000)

# Right-Skewed (Positive Skew) - using an exponential distribution
data_right_skewed = np.random.exponential(scale=1, size=1000)

# Left-Skewed (Negative Skew) - mirroring the exponential distribution
data_left_skewed = -np.random.exponential(scale=1, size=1000)

# 3. Set up a 3x2 grid of plots
fig, axes = plt.subplots(nrows=5, ncols=2, figsize=(12, 14))
fig.suptitle('Q-Q Plots: Peaked (Heavy-Tailed) vs. Flat (Light-Tailed)', fontsize=16, y=1.02)

def plot_distribution(data, title, row, color):
    """Helper function to plot a histogram and its corresponding Q-Q plot."""
    # Plot Histogram
    sns.histplot(data, kde=True, ax=axes[row, 0], color=color, stat='density')
    axes[row, 0].set_title(f"{title} - Histogram", fontsize=12)
    axes[row, 0].set_xlabel("Value")
    axes[row, 0].set_ylabel("Density")

    # Plot Q-Q Plot
    stats.probplot(data, dist="norm", plot=axes[row, 1])
    axes[row, 1].set_title(f"{title} - Q-Q Plot", fontsize=12)
    axes[row, 1].get_lines()[0].set_markerfacecolor(color)
    axes[row, 1].get_lines()[0].set_markeredgecolor(color)
    axes[row, 1].get_lines()[0].set_alpha(0.5)

# 4. Generate the plots
plot_distribution(data_normal, "Normal Distribution", 0, "dodgerblue")
plot_distribution(data_peaked, "Peaked / Heavy-Tailed (Laplace)", 1, "mediumorchid")
plot_distribution(data_flat, "Flat / Light-Tailed (Uniform)", 2, "darkorange")
plot_distribution(data_right_skewed, "Right-Skewed Distribution", 3, "coral")
plot_distribution(data_left_skewed, "Left-Skewed Distribution", 4, "mediumseagreen")

# 5. Adjust layout and display
plt.tight_layout()
plt.savefig('qq_plot_distributions.png', dpi=300, bbox_inches='tight')
plt.show()
```

Let's walk through the exact mathematical steps to calculate a percentile by hand, using the **linear interpolation** method (which is the default used by NumPy and most statistical software).

We will calculate the **75th percentile** for a small, random dataset.

### Step 1: Define and Sort the Data

Imagine you have the following raw data points:
Data: $[45, 12, 89, 23, 78, 56]$

The very first step is always to sort the data in ascending order. We will also assign a 0-based index to each value (just like Python does):

| Index ($i$) | 0 | 1 | 2 | 3 | 4 | 5 |
| --- | --- | --- | --- | --- | --- | --- |
| **Value** | 12 | 23 | 45 | 56 | 78 | 89 |

Here, our total number of elements is $N = 6$.

### Step 2: Calculate the Rank (Index)

Next, we calculate the exact fractional index (or rank) where the 75th percentile sits.

The formula for the target index $i$ is:


$$i = \frac{q}{100} \times (N - 1)$$

Where:

* $q$ is our target percentile ($75$)
* $N$ is the total number of elements ($6$)

Plugging in our numbers:


$$i = \frac{75}{100} \times (6 - 1)$$

$$i = 0.75 \times 5$$

$$i = 3.75$$

This tells us the 75th percentile sits at exactly **index 3.75**—which is $75\%$ of the way between the value at index `3` and the value at index `4`.

### Step 3: Split the Rank

To do the interpolation, we split that calculated index ($3.75$) into two parts:

* **The integer part ($k$):** $3$ (This points to our lower bound)
* **The fractional part ($d$):** $0.75$ (This is our multiplier)

Looking back at our sorted table:

* Value at index $3$ ($V_k$) = $56$
* Value at index $4$ ($V_{k+1}$) = $78$

### Step 4: Linear Interpolation

Now we find the exact value. We take the lower value and add the fractional difference between the upper and lower values.

The formula is:


$$\text{Percentile Value} = V_k + d \times (V_{k+1} - V_k)$$

Plugging in our numbers:


$$\text{Percentile Value} = 56 + 0.75 \times (78 - 56)$$

$$\text{Percentile Value} = 56 + 0.75 \times (22)$$

$$\text{Percentile Value} = 56 + 16.5$$

$$\text{Percentile Value} = 72.5$$

### The Final Result

The 75th percentile of our dataset is mathematically exactly **$72.5$**.

Because $72.5$ sits precisely $75\%$ of the way between $56$ and $78$ on a number line, this method provides a much more accurate representation of continuous data than simply rounding to the nearest existing data point!
